# 🤖 NLP Buddy – Web Application ระบบ RAG

**หัวข้อที่เลือก:** ผู้ช่วยตอบคำถามด้านการประมวลผลภาษาธรรมชาติ (NLP) ด้วย Retrieval-Augmented Generation

- **URL หน้าเว็บ Streamlit:** https://nlp-buddy-rag.streamlit.app
- **GitHub Repository URL:** https://github.com/Namenick-2004/-NLP

## แนวคิดของระบบ

NLP Buddy รวบรวมเอกสารความรู้ภาษาไทยด้าน NLP จำนวน 12 หัวข้อ ผู้ใช้สามารถถามเป็นภาษาธรรมชาติ ระบบจะค้นหา Chunk ที่เกี่ยวข้องด้วย Sentence Embedding และ FAISS ก่อนส่งบริบทให้ Gemini เรียบเรียงคำตอบ พร้อมแสดงเอกสารอ้างอิงและคะแนนความคล้าย

## RAG Pipeline

1. **Document Loading & Cleaning** – โหลดไฟล์ Markdown/Text และทำความสะอาดข้อความ
2. **Chunking** – แบ่งเอกสารเป็น Chunk ขนาดไม่เกิน 500 ตัวอักษร พร้อม Overlap
3. **Embedding** – ใช้ paraphrase-multilingual-MiniLM-L12-v2
4. **Vector Search** – ค้น Top-4 ด้วย FAISS และ Cosine Similarity
5. **Prompt Engineering** – บังคับให้ตอบจากบริบทและอ้างอิง [1], [2]
6. **Large Language Model** – ใช้ Gemini สร้างคำตอบ
7. **Guardrail** – ตอบว่าไม่พบข้อมูลเมื่อคะแนนต่ำกว่าเกณฑ์หรือคำถามไม่ตรงกับเอกสาร
